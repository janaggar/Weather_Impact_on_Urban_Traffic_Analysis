```python
import pandas as pd

# Load the raw data
weather = pd.read_csv('weather_raw.csv')

# --- Step 2: Data Cleaning (Handling Missing Values) ---

# Calculate imputation values
city_mode = weather['city'].mode()[0]
weather_mode = weather['weather_condition'].mode()[0]
temp_median = weather['temperature_c'].median()

# Impute missing values using direct assignment to avoid FutureWarning
weather['city'] = weather['city'].fillna(city_mode)
weather['weather_condition'] = weather['weather_condition'].fillna(weather_mode)
weather['season'] = weather['season'].fillna('Unknown')
weather['temperature_c'] = weather['temperature_c'].fillna(temp_median)


# --- Step 3: Date Type Conversion and Formatting ---

# Convert 'date_time' to datetime objects
weather['date_time'] = pd.to_datetime(weather['date_time'], errors='coerce')

# Drop rows where date_time conversion might have failed (safety step)
weather.dropna(subset=['date_time'], inplace=True)

# Reformat 'date_time' to the requested 'DD/MM/YYYY HH:MM' string format
weather['date_time'] = weather['date_time'].dt.strftime('%d/%m/%Y %H:%M')

# --- Step 4: Column Reordering (Making 'date_time' the first column) ---

cols = weather.columns.tolist()
cols.insert(0, cols.pop(cols.index('date_time')))
weather = weather[cols]

# Save the final cleaned and formatted data
weather.to_csv('weather_cleaned.csv', index=False)
```
