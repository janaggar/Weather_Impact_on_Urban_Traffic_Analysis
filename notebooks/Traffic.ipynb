```python
import pandas as pd

# Load the raw data
traffic = pd.read_csv('traffic_raw.csv')

# Calculate modes for all four categorical columns with nulls
city_mode = traffic['city'].mode()[0]
area_mode = traffic['area'].mode()[0]
congestion_mode = traffic['congestion_level'].mode()[0]
road_mode = traffic['road_condition'].mode()[0]

# Impute missing values using direct assignment to ensure correctness
traffic['city'] = traffic['city'].fillna(city_mode)
traffic['area'] = traffic['area'].fillna(area_mode)
traffic['congestion_level'] = traffic['congestion_level'].fillna(congestion_mode)
traffic['road_condition'] = traffic['road_condition'].fillna(road_mode)

# Convert 'date_time' to datetime objects (handling mixed formats)
traffic['date_time'] = pd.to_datetime(traffic['date_time'], errors='coerce')

# Drop rows where date_time conversion might have failed (safety step)
# Note: For this dataset, no rows are expected to be dropped.
traffic.dropna(subset=['date_time'], inplace=True)

# Reformat 'date_time' to the consistent 'DD/MM/YYYY HH:MM' string format
traffic['date_time'] = traffic['date_time'].dt.strftime('%d/%m/%Y %H:%M')

# Get the list of all column names
cols = traffic.columns.tolist()

# Move 'date_time' to the first position (index 0)
cols.insert(0, cols.pop(cols.index('date_time')))

# Apply the new column order to the DataFrame
traffic = traffic[cols]

# Save the final cleaned and formatted data
traffic.to_csv('traffic_cleaned.csv', index=False)
```
